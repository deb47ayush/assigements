```python
#PS To find the largest of the 3 numbers
a = int(input("Please enter the 1st num:"))
b = int(input("Please enter the 2nd num:"))
c = int(input("Please enter the 3rd num:"))

if (a>b) and (a>c):
    print("a is the greatest num")
elif (b>a) and (b>c):
    print("b is the greatest num")
elif (c>a) and (c>b):
    print("c is the greatest num")
else:
    print("All num are equal")
```

```python
#PS To check wheather it is right angled or not using function

a = int(input("Please enter 1st side"))
b = int(input("Please enter 2nd side"))
c = int(input("Please enter 3rd side"))

def right(a,b,c):
    if (a*a == b*b + c*c):
        print("Right angled triangle")
    elif (b*b == c*c + a*a):
        print("Right angled triangle")
    elif (c*c == a*a + b*b):
        print("Right angled triangle")
    else:
        print("not Right angled triangle")

right(a,b,c)
```
